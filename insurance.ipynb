Supervised learning
Classification type - Binay classification

Predict whether isurance is taken or not

```python
import pandas as pd
```

```python
df = pd.read_csv("insurance.csv")
```

```python
df.head()
```

```python
x = df[['age']]
```

```python
y = df[['insurance']]
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(x,y)
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
model.predict([[50]])
```

```python
Age = int(input("Enter the age : "))
res = model.predict([[Age]])
if res[0]==1:
    print("Insurace will be opted")
else:
    print("Insurance will not be opted")
```
