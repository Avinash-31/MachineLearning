```python
import pandas as pd
```

```python
# read data
diabetes = pd.read_csv('https://github.com/YBIFoundation/Dataset/raw/main/Diabetes.csv')
```

```python
diabetes.head()
```

```python
diabetes.columns
```

```python
y = diabetes['diabetes']
```

```python
x = diabetes.drop(['diabetes'],axis = 1)
```

```python
x = diabetes[['pregnancies', 'glucose', 'diastolic', 'triceps', 'insulin', 'bmi',
       'dpf', 'age']]
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_x,test_x,train_y,test_y = train_test_split(x,y,test_size = 0.2,random_state = 2529)
```

```python
train_x.shape,train_y.shape,test_x.shape,test_y.shape
```

```python
train_x
```

```python
# model 1
from sklearn.linear_model import LogisticRegression
model1 = LogisticRegression(max_iter = 500)
model1.fit(train_x,train_y)
```

```python
# model 2
from sklearn.tree import DecisionTreeClassifier
model2 = DecisionTreeClassifier()
model2.fit(train_x,train_y)
```

```python
# model 3
from sklearn.neighbors import KNeighborsClassifier
model3 = KNeighborsClassifier(n_neighbors = 10)
model3.fit(train_x,train_y)
```

```python
model1_pred = model1.predict(test_x)
model2_pred = model2.predict(test_x)
model3_pred = model3.predict(test_x)
```

```python
model1.score(test_x,test_y)
```

```python
# Step 8 : model accuracy
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
accuracy_score(test_y,model1_pred)
```

```python
accuracy_score(test_y,model2_pred)
```

```python
accuracy_score(test_y,model3_pred)
```
