```python
maxi = 0
n = 0
for i in range(1,50):
    model = KNeighborsClassifier(n_neighbors = i)
    model.fit(train_x,train_y)
    score = model.score(test_x,test_y)
    if(score>maxi):
        maxi = score
        n=i
print(f"Maximum score is achived when neighbors are {n}")
```
