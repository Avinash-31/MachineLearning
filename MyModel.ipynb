```python
import pandas as pd
```

```python
df = pd.read_csv("areaPrice.csv")
```

```python
df
```

```python
x = df[['area']]
```

```python
y = df.price
```

```python
pip install scikit-learn
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x.values,y.values)
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
model.predict([[1000]])
```

```python
area = int(input("Enter the value of area "))
res = model.predict([[area]])
print(f"Price of the area {area} is {res[0]}")
```
