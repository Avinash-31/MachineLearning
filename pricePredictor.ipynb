Price predictor
Price varies on age, number of bedrooms, age and area
(Multi linear regression)

```python
import pandas as pd
```

```python
df = pd.read_csv("multi.csv")
```

```python
df
```

```python
x = df[["area","bedrooms","age"]]
```

```python
x
```

```python
y = df[["price"]]
```

```python
y
```

```python
import matplotlib.pyplot as plt
plt.title("Price vs other parameters")
plt.xlabel("Other parameters")
plt.ylabel("Price")
plt.plot(x,y)
plt.legend(['Area','Age','Number of bedrooms'])
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x,y)
```

m1,m2,m3 values

```python
model.coef_
```

```python
#intercept values
model.intercept_
```

```python
model.predict([[5000,3,5]])
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
Area = int(input("Enter the area of the building"))
Age = int(input("Enter the age of the building"))
Bedrooms = int(input("Enter the number of bedrooms in the buulding"))
res = model.predict([[Area,Age,Bedrooms]])
print(f"Price of the building : {res[0][0}")
```
